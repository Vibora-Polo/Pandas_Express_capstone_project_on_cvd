### PROBLEM STATEMENT


Cardiovascular disease—including heart disease and heart attack—is a growing public health crisis in Kenya, driven by rapid urbanization, dietary shifts toward processed foods, rising obesity, increasing tobacco use, and limited access to preventive screening in many communities. Kenya's Ministry of Health and county health departments need reliable, interpretable signals to identify high-risk populations and allocate scarce prevention resources; healthcare providers in Kenyan hospitals, clinics, and community health units need simple, evidence-based tools to support opportunistic screening during routine visits; health systems such as NHIF/SHA and private insurers need data-driven approaches to design preventive wellness programs and reduce the growing burden of non-communicable disease (NCD) admissions; and Kenyan patients and community health volunteers (CHVs) need understandable insights that translate into actionable lifestyle changes within the local context.

While Kenya does not yet have a population-scale behavioral risk factor survey comparable in size and variable richness to the U.S. CDC BRFSS, the risk factors captured by BRFSS—smoking, obesity/BMI, physical inactivity, diabetes, hypertension, high cholesterol, poor general health, and inadequate sleep—are the same modifiable risk factors driving cardiovascular disease in Kenya (as documented by Kenya's National NCD Strategic Plan and the WHO STEPwise surveys). This project therefore uses the CDC BRFSS 2024 raw public dataset as a methodological proxy and proof-of-concept: the goal is to develop and validate an interpretable machine learning pipeline and deployable screening tool that could be adapted for Kenyan populations once equivalent local survey data (e.g., from the Kenya STEPwise survey, Kenya Demographic and Health Survey, or a future Kenya BRFSS-style instrument) becomes available at sufficient scale.

Build and evaluate an interpretable machine learning model using CDC BRFSS 2024 data that estimates an individual's risk of having heart disease or a prior heart attack (combined outcome) based on behavioral and clinical self-report indicators.

Deploy a screening-style web application where a user can input risk factors and receive a risk probability, a high/low risk classification, and an explanation of the top contributing factors—designed so that the interface, variable set, and explanation approach could be directly reused with Kenyan survey data when available.
Discuss transferability: explicitly analyze which BRFSS-derived findings are likely to generalize to Kenyan populations and which may not, citing differences in demographics, healthcare access, dietary patterns, and genetic/environmental context.

# Business Understanding

## 1. Introduction: The Real-World Problem

Cardiovascular disease (CVD)—including coronary heart disease and heart attack (myocardial infarction)—is the **leading cause of death globally** and a rapidly growing crisis in Kenya. According to the World Health Organization and Kenya's National NCD Strategic Plan, non-communicable diseases now account for a significant and rising share of mortality and morbidity in Kenya, with cardiovascular conditions at the forefront. The key drivers are **modifiable behavioral and clinical risk factors**: tobacco use, physical inactivity, unhealthy diets leading to obesity, diabetes, hypertension, and high cholesterol.

### Why this matters for Kenya specifically

- **Rapid urbanization** is shifting dietary patterns toward processed, high-sodium, high-sugar foods.
- **Tobacco use** remains prevalent, particularly among men and in certain regions.
- **Limited access to preventive screening**: most Kenyans interact with the health system only when acutely ill; opportunistic CVD risk screening during routine visits is rare, especially at the community health unit level.
- **Resource constraints**: Kenya's 47 county health departments must allocate scarce NCD budgets efficiently—targeting the populations and risk factors that matter most.
- **Growing insurance coverage** (NHIF/SHA transition) creates both an opportunity and a need for data-driven preventive wellness program design.

### The gap this project addresses

There is no widely available, simple, data-driven cardiovascular risk screening tool designed for the Kenyan health context that:
1. uses **readily collectible behavioral and clinical self-report indicators** (no lab tests required),
2. provides an **interpretable risk score** (not a black-box prediction),
3. **explains which factors** are driving an individual's risk (actionable for behavior change), and
4. is **deployable** as a lightweight web tool usable by community health volunteers, clinics, and public health planners.

### Why BRFSS as a proxy dataset

While Kenya does not yet have a population-scale behavioral risk factor survey of sufficient size and variable richness for machine learning, the **CDC BRFSS 2024** dataset captures the **same modifiable risk factors** driving CVD in Kenya—smoking, BMI/obesity, physical inactivity, diabetes, hypertension, high cholesterol, general health status, and sleep. This project uses BRFSS as a **methodological proof-of-concept**: the pipeline, model, and deployed tool are designed so they can be **directly reused with Kenyan survey data** (e.g., Kenya STEPwise survey, KDHS) once available at sufficient scale.

> ⚠️ **Important disclaimer:** This project is built on U.S. survey data and is intended for research, education, and proof-of-concept purposes only. It is not validated for clinical use in Kenya and should not replace medical judgment.

### Transferability to Kenyan data

This project was built on U.S. BRFSS data as a **proof-of-concept**. The pipeline—data ingestion, feature engineering, modeling, evaluation, and deployment—is designed to be **directly reusable** with Kenyan survey data (e.g., Kenya STEPwise survey, KDHS) once available at sufficient scale. Key considerations for adaptation:
- **Likely transferable:** the direction and relative importance of risk factors (smoking, obesity, inactivity, diabetes, hypertension) are consistent across populations.
- **Requires recalibration:** absolute risk probabilities will differ due to differences in demographics, genetic background, dietary patterns, healthcare access, and disease prevalence.
- **Requires local validation:** any deployment in a Kenyan clinical or public health setting must be validated against local outcome data before operational use.

### Limitations
- **Cross-sectional data:** the model classifies current risk status, not future onset.
- **Self-report bias:** all variables are self-reported; under/over-reporting is possible.
- **U.S. population:** findings may not directly generalize to Kenyan populations without recalibration.
- **Not a diagnostic tool:** this is a screening aid and educational proof-of-concept, not a replacement for clinical evaluation.
